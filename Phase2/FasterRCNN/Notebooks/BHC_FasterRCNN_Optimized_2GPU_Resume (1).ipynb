```python
from PIL import Image
import torch
from torch.utils.data import Dataset
import torchvision.transforms.functional as F

RAW_TO_MODEL = {0: 1, 1: 2}   # torchvision reserves label 0 for background
MODEL_TO_RAW = {1: 0, 2: 1}
NUM_CLASSES_WITH_BG = 3

class YoloDetectionDataset(Dataset):
    def __init__(self, split_dir, transforms=None):
        self.images_dir = os.path.join(split_dir, "images")
        self.labels_dir = os.path.join(split_dir, "labels")
        self.transforms = transforms
        self.image_files = sorted([
            f for f in os.listdir(self.images_dir)
            if f.lower().endswith(('.jpg', '.jpeg', '.png'))
        ])

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img_name = self.image_files[idx]
        img_path = os.path.join(self.images_dir, img_name)
        label_path = os.path.join(self.labels_dir, os.path.splitext(img_name)[0] + ".txt")

        img = Image.open(img_path).convert("RGB")
        img_w, img_h = img.size

        boxes, labels = [], []
        if os.path.exists(label_path):
            with open(label_path) as f:
                for line in f:
                    parts = line.strip().split()
                    if len(parts) != 5:
                        continue
                    cls, xc, yc, w, h = parts
                    cls = int(cls)
                    xc, yc, w, h = float(xc), float(yc), float(w), float(h)

                    x1 = (xc - w / 2) * img_w
                    y1 = (yc - h / 2) * img_h
                    x2 = (xc + w / 2) * img_w
                    y2 = (yc + h / 2) * img_h

                    if x2 <= x1 or y2 <= y1:
                        continue

                    boxes.append([x1, y1, x2, y2])
                    labels.append(RAW_TO_MODEL[cls])

        boxes = torch.as_tensor(boxes, dtype=torch.float32) if boxes else torch.zeros((0, 4), dtype=torch.float32)
        labels = torch.as_tensor(labels, dtype=torch.int64) if labels else torch.zeros((0,), dtype=torch.int64)

        target = {
            "boxes": boxes,
            "labels": labels,
            "image_id": torch.tensor([idx]),
            "area": (boxes[:, 3] - boxes[:, 1]) * (boxes[:, 2] - boxes[:, 0]) if len(boxes) else torch.zeros((0,)),
            "iscrowd": torch.zeros((len(boxes),), dtype=torch.int64),
        }

        img = F.to_tensor(img)
        if self.transforms:
            img, target = self.transforms(img, target)
        return img, target

def collate_fn(batch):
    return tuple(zip(*batch))
```

```python
import torchvision
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor

def get_model(num_classes=NUM_CLASSES_WITH_BG):
    model = torchvision.models.detection.fasterrcnn_resnet50_fpn_v2(weights="DEFAULT")
    in_features = model.roi_heads.box_predictor.cls_score.in_features
    model.roi_heads.box_predictor = FastRCNNPredictor(in_features, num_classes)
    return model

model = get_model()
model.to(device)
```

```python
from torch.utils.data import DataLoader
import os
import torch

TRAIN_DIR = os.path.join(LOCAL_DATASET_PATH, "train")
VALID_DIR = os.path.join(LOCAL_DATASET_PATH, "valid")
TEST_DIR  = os.path.join(LOCAL_DATASET_PATH, "test")

train_dataset = YoloDetectionDataset(TRAIN_DIR)
valid_dataset = YoloDetectionDataset(VALID_DIR)
test_dataset  = YoloDetectionDataset(TEST_DIR)

print("Train:", len(train_dataset), " Valid:", len(valid_dataset), " Test:", len(test_dataset))

# Batch size is split across the GPUs by DataParallel.
# With 2 GPUs, batch size 4 means roughly 2 images per GPU.
BATCH_SIZE = 4
NUM_WORKERS = min(4, os.cpu_count() or 2)
PIN_MEMORY = torch.cuda.is_available()

train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=NUM_WORKERS, collate_fn=collate_fn,
    pin_memory=PIN_MEMORY, persistent_workers=NUM_WORKERS > 0
)
valid_loader = DataLoader(
    valid_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, collate_fn=collate_fn,
    pin_memory=PIN_MEMORY, persistent_workers=NUM_WORKERS > 0
)
test_loader = DataLoader(
    test_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, collate_fn=collate_fn,
    pin_memory=PIN_MEMORY, persistent_workers=NUM_WORKERS > 0
)

img, target = train_dataset[0]
print("Sample image shape:", img.shape, "target keys:", list(target.keys()))
```

```python
import os
import time
import json
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

# ============================================================
# OPTIMIZED TRAINING: multi-GPU DataParallel + AMP + checkpoint
# ============================================================

NUM_EPOCHS = 150
CKPT_PATH = os.path.join(DATASET_DRIVE_PATH, "fasterrcnn_resume.pth")
MODEL_PATH = os.path.join(DATASET_DRIVE_PATH, "fasterrcnn_beach_human.pth")
HISTORY_PATH = os.path.join(DATASET_DRIVE_PATH, "history.json")
LOSS_GRAPH_PATH = os.path.join(DATASET_DRIVE_PATH, "loss_curve.png")
COMPONENT_GRAPH_PATH = os.path.join(DATASET_DRIVE_PATH, "loss_components.png")

device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
num_gpus = torch.cuda.device_count()
print("Device:", device, "| GPU count:", num_gpus)
for i in range(num_gpus):
    print(f"GPU {i}: {torch.cuda.get_device_name(i)}")

# If an earlier cell wrapped model in DataParallel, unwrap it before re-wrapping.
if isinstance(model, nn.DataParallel):
    model = model.module
model = model.to(device)

# Torchvision detection returns a dict of scalar losses. DataParallel cannot
# reliably gather scalar-valued dicts, so this wrapper returns a tuple instead.
class DetectionLossWrapper(nn.Module):
    def __init__(self, detector):
        super().__init__()
        self.detector = detector

    def forward(self, images, targets):
        loss_dict = self.detector(images, targets)
        return (
            loss_dict["loss_classifier"],
            loss_dict["loss_box_reg"],
            loss_dict["loss_objectness"],
            loss_dict["loss_rpn_box_reg"],
        )

use_multi_gpu = torch.cuda.is_available() and num_gpus >= 2
if use_multi_gpu:
    train_model = nn.DataParallel(DetectionLossWrapper(model), device_ids=list(range(num_gpus)))
    print(f"Training across {num_gpus} GPUs with DataParallel")
else:
    train_model = DetectionLossWrapper(model)
    print("Training on one GPU or CPU")

params = [p for p in model.parameters() if p.requires_grad]
optimizer = torch.optim.SGD(params, lr=0.005, momentum=0.9, weight_decay=0.0005)
lr_scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=4, gamma=0.1)
scaler = torch.amp.GradScaler("cuda", enabled=(device.type == "cuda"))

history = {
    "train_loss": [], "val_loss": [],
    "loss_classifier": [], "loss_box_reg": [],
    "loss_objectness": [], "loss_rpn_box_reg": []
}
start_epoch = 0

# Resume only from checkpoints created by this cell.
if os.path.exists(CKPT_PATH):
    checkpoint = torch.load(CKPT_PATH, map_location=device, weights_only=False)
    model.load_state_dict(checkpoint["model_state_dict"])
    optimizer.load_state_dict(checkpoint["optimizer_state_dict"])
    lr_scheduler.load_state_dict(checkpoint["scheduler_state_dict"])
    if "scaler_state_dict" in checkpoint:
        scaler.load_state_dict(checkpoint["scaler_state_dict"])
    history = checkpoint.get("history", history)
    start_epoch = int(checkpoint["epoch"])
    print(f"Resuming after epoch {start_epoch}; next epoch is {start_epoch + 1}")
else:
    print("No checkpoint found. Starting from pretrained COCO weights at epoch 1.")

def evaluate_loss(detector, data_loader):
    # Detection models calculate loss dictionaries only in train mode.
    detector.train()
    total_loss, count = 0.0, 0
    with torch.no_grad():
        for images, targets in data_loader:
            images = [img.to(device, non_blocking=True) for img in images]
            targets = [{k: v.to(device, non_blocking=True) for k, v in t.items()} for t in targets]
            with torch.autocast(device_type=device.type, dtype=torch.float16,
                                enabled=(device.type == "cuda")):
                loss_dict = detector(images, targets)
                loss = sum(loss_dict.values())
            total_loss += loss.item()
            count += 1
    return total_loss / max(count, 1)

for epoch in range(start_epoch, NUM_EPOCHS):
    model.train()
    train_model.train()
    total_loss = 0.0
    components = {
        "loss_classifier": 0.0, "loss_box_reg": 0.0,
        "loss_objectness": 0.0, "loss_rpn_box_reg": 0.0
    }
    t0 = time.time()

    for images, targets in train_loader:
        images = [img.to(device, non_blocking=True) for img in images]
        targets = [{k: v.to(device, non_blocking=True) for k, v in t.items()} for t in targets]

        optimizer.zero_grad(set_to_none=True)
        with torch.autocast(device_type=device.type, dtype=torch.float16,
                            enabled=(device.type == "cuda")):
            outputs = train_model(images, targets)
            # DataParallel gathers each scalar loss into a vector (one value per GPU).
            loss_values = [value.mean() for value in outputs]
            losses = sum(loss_values)

        if not torch.isfinite(losses):
            raise RuntimeError(f"Non-finite loss at epoch {epoch + 1}: {losses.item()}")

        scaler.scale(losses).backward()
        scaler.step(optimizer)
        scaler.update()

        total_loss += losses.detach().item()
        for key, value in zip(components.keys(), loss_values):
            components[key] += value.detach().item()

    avg_train_loss = total_loss / max(len(train_loader), 1)
    avg_val_loss = evaluate_loss(model, valid_loader)
    lr_scheduler.step()

    history["train_loss"].append(avg_train_loss)
    history["val_loss"].append(avg_val_loss)
    for key in components:
        history[key].append(components[key] / max(len(train_loader), 1))

    completed_epoch = epoch + 1
    elapsed = time.time() - t0
    print(
        f"Epoch [{completed_epoch}/{NUM_EPOCHS}] "
        f"train_loss={avg_train_loss:.4f} "
        f"val_loss={avg_val_loss:.4f} "
        f"time={elapsed/60:.1f} min"
    )

    # Atomically save full resume state after every epoch.
    state = {
        "epoch": completed_epoch,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "scheduler_state_dict": lr_scheduler.state_dict(),
        "scaler_state_dict": scaler.state_dict(),
        "history": history
    }
    temp_path = CKPT_PATH + ".tmp"
    torch.save(state, temp_path)
    os.replace(temp_path, CKPT_PATH)
    torch.save(model.state_dict(), MODEL_PATH)
    with open(HISTORY_PATH, "w") as f:
        json.dump(history, f)

    # Update graphs after each epoch.
    plt.figure(figsize=(9, 5))
    plt.plot(history["train_loss"], label="Training loss")
    plt.plot(history["val_loss"], label="Validation loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("Faster R-CNN Training and Validation Loss")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(LOSS_GRAPH_PATH, dpi=180)
    plt.close()

    plt.figure(figsize=(10, 6))
    for key in ["loss_classifier", "loss_box_reg", "loss_objectness", "loss_rpn_box_reg"]:
        plt.plot(history[key], label=key)
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("Faster R-CNN Loss Components")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(COMPONENT_GRAPH_PATH, dpi=180)
    plt.close()

print("Training loop finished.")
print("Model:", MODEL_PATH)
print("Checkpoint:", CKPT_PATH)
print("Loss graph:", LOSS_GRAPH_PATH)
print("Loss components:", COMPONENT_GRAPH_PATH)
```

```python
model.eval()
all_preds, all_targets = [], []

with torch.no_grad():
    for images, targets in test_loader:
        images = [img.to(device) for img in images]
        outputs = model(images)
        preds = [{k: v.cpu() for k, v in o.items()} for o in outputs]
        tgts  = [{k: v.cpu() for k, v in t.items()} for t in targets]
        all_preds.extend(preds)
        all_targets.extend(tgts)

print("Collected predictions for", len(all_preds), "test images")
```

```python
from torchmetrics.detection.mean_ap import MeanAveragePrecision

metric = MeanAveragePrecision(class_metrics=True)
metric.update(all_preds, all_targets)
results = metric.compute()

print("mAP@[0.5:0.95]:", round(results['map'].item(), 4))
print("mAP@0.5:        ", round(results['map_50'].item(), 4))
print("mAP@0.75:       ", round(results['map_75'].item(), 4))
print("mAR (100 dets): ", round(results['mar_100'].item(), 4))

if 'map_per_class' in results and results['map_per_class'].numel() > 0:
    classes = results.get('classes', torch.tensor([1, 2])).tolist()
    for model_label, ap in zip(classes, results['map_per_class'].tolist()):
        raw = MODEL_TO_RAW.get(model_label, model_label)
        print(f"  AP class {raw} ({CLASS_NAMES_RAW.get(raw, raw)}): {round(ap,4)}")
```

```python
import numpy as np

def box_iou(box1, box2):
    x1, y1 = max(box1[0], box2[0]), max(box1[1], box2[1])
    x2, y2 = min(box1[2], box2[2]), min(box1[3], box2[3])
    inter = max(0, x2-x1) * max(0, y2-y1)
    area1 = (box1[2]-box1[0]) * (box1[3]-box1[1])
    area2 = (box2[2]-box2[0]) * (box2[3]-box2[1])
    union = area1 + area2 - inter
    return inter / union if union > 0 else 0

IOU_THRESH = 0.5
SCORE_THRESH = 0.5

y_true, y_pred, y_scores = [], [], []
total_gt, total_matched, total_fp = 0, 0, 0

for pred, tgt in zip(all_preds, all_targets):
    matched_gt = set()
    p_boxes, p_labels, p_scores = pred['boxes'].numpy(), pred['labels'].numpy(), pred['scores'].numpy()
    g_boxes, g_labels = tgt['boxes'].numpy(), tgt['labels'].numpy()
    total_gt += len(g_boxes)

    keep = p_scores > SCORE_THRESH
    p_boxes, p_labels, p_scores = p_boxes[keep], p_labels[keep], p_scores[keep]

    for pb, pl, ps in zip(p_boxes, p_labels, p_scores):
        best_iou, best_j = 0, -1
        for j, gb in enumerate(g_boxes):
            if j in matched_gt:
                continue
            iou = box_iou(pb, gb)
            if iou > best_iou:
                best_iou, best_j = iou, j
        if best_iou >= IOU_THRESH:
            matched_gt.add(best_j)
            total_matched += 1
            y_true.append(MODEL_TO_RAW.get(int(g_labels[best_j]), int(g_labels[best_j])))
            y_pred.append(MODEL_TO_RAW.get(int(pl), int(pl)))
            y_scores.append(ps)
        else:
            total_fp += 1  # unmatched prediction = false positive

y_true, y_pred, y_scores = np.array(y_true), np.array(y_pred), np.array(y_scores)
print(f"Total GT boxes: {total_gt} | Matched (TP): {total_matched} | Unmatched preds (FP): {total_fp}")
```

```python
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                              classification_report, confusion_matrix, ConfusionMatrixDisplay)

detection_accuracy = total_matched / total_gt if total_gt > 0 else 0
classification_accuracy = accuracy_score(y_true, y_pred)  # accuracy of 0-vs-1 label among matched detections

precision_macro = precision_score(y_true, y_pred, average='macro', zero_division=0)
recall_macro = recall_score(y_true, y_pred, average='macro', zero_division=0)
f1_macro = f1_score(y_true, y_pred, average='macro', zero_division=0)

print(f"Detection recall (TP / total GT boxes): {detection_accuracy:.4f}")
print(f"Classification accuracy (0 vs 1, on matched boxes): {classification_accuracy:.4f}")
print(f"Precision (macro): {precision_macro:.4f}")
print(f"Recall (macro):    {recall_macro:.4f}")
print(f"F1-score (macro):  {f1_macro:.4f}")
print()
print(classification_report(y_true, y_pred, target_names=[CLASS_NAMES_RAW.get(0,"class0"), CLASS_NAMES_RAW.get(1,"class1")], zero_division=0))
```

```python
import matplotlib.patches as patches

def show_predictions(dataset, model, num_images=4, score_thresh=0.6):
    model.eval()
    fig, axes = plt.subplots(1, num_images, figsize=(5*num_images, 5))
    for i in range(num_images):
        img, target = dataset[i]
        with torch.no_grad():
            pred = model([img.to(device)])[0]
        img_np = img.permute(1,2,0).numpy()
        ax = axes[i] if num_images > 1 else axes
        ax.imshow(img_np)
        for box, label, score in zip(pred['boxes'], pred['labels'], pred['scores']):
            if score < score_thresh:
                continue
            raw_label = MODEL_TO_RAW.get(int(label), int(label))
            x1,y1,x2,y2 = box.cpu().numpy()
            rect = patches.Rectangle((x1,y1), x2-x1, y2-y1, linewidth=2, edgecolor='lime', facecolor='none')
            ax.add_patch(rect)
            ax.text(x1, y1-5, f"{raw_label}:{score:.2f}", color='lime', fontsize=10, weight='bold')
        ax.axis('off')
    plt.tight_layout()
    plt.savefig("/content/sample_predictions.png", dpi=150, bbox_inches='tight')
    plt.show()

show_predictions(test_dataset, model, num_images=4)
```

```python
# ============================================================
# TEST EXACTLY 10 AVAILABLE TEST IMAGES
# AUTOMATICALLY REPLACES MISSING REQUESTED IMAGES
#
# 0 = LAND
# 1 = WATER
# ============================================================

import os
import shutil
import numpy as np
import pandas as pd
import torch
import torchvision
import matplotlib.pyplot as plt
import matplotlib.patches as patches

from PIL import Image
from torchvision.transforms import functional as F
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    precision_recall_curve,
    average_precision_score
)

# ============================================================
# 1. PATHS
# ============================================================

DATASET_PATH = "LOCAL_DATASET_PATH"

TEST_IMAGE_DIR = os.path.join(
    DATASET_PATH,
    "test",
    "images"
)

TEST_LABEL_DIR = os.path.join(
    DATASET_PATH,
    "test",
    "labels"
)

CLEAN_MODEL = os.path.join(
    DATASET_PATH,
    "fasterrcnn_beach_human_clean.pth"
)

ORIGINAL_MODEL = os.path.join(
    DATASET_PATH,
    "fasterrcnn_beach_human.pth"
)

# ============================================================
# 2. MODEL
# ============================================================

if os.path.exists(CLEAN_MODEL):
    MODEL_PATH = CLEAN_MODEL
else:
    MODEL_PATH = ORIGINAL_MODEL

print("Model:")
print(MODEL_PATH)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Using device:", device)

NUM_CLASSES = 3

model = torchvision.models.detection.fasterrcnn_resnet50_fpn_v2(
    weights=None
)

in_features = (
    model.roi_heads
    .box_predictor
    .cls_score
    .in_features
)

model.roi_heads.box_predictor = FastRCNNPredictor(
    in_features,
    NUM_CLASSES
)

checkpoint = torch.load(
    MODEL_PATH,
    map_location=device
)

model.load_state_dict(checkpoint)

model.to(device)
model.eval()

print("✅ Model loaded successfully!")

# ============================================================
# 3. SETTINGS
# ============================================================

CONFIDENCE_THRESHOLD = 0.50
IOU_THRESHOLD = 0.50

MODEL_TO_RAW = {
    1: 0,   # model class 1 -> land
    2: 1    # model class 2 -> water
}

# ============================================================
# 4. YOUR REQUESTED IMAGES
# ============================================================

requested_images = [
    "1640154603-Wed-Dec-22_16_30_03-AEST-2021-BURLEIGH-TWR-snap-4Gtrailcam_jpg.rf.JRKPYKB0XT4cxKdpZm3b.jpg",

    "1640150102-Wed-Dec-22_15_15_02-AEST-2021-BURLEIGH-TWR-snap-4Gtrailcam_jpg.rf.WSuIM7YlMnrO43HvoTSi.jpg",

    "1640151902-Wed-Dec-22_15_45_02-AEST-2021-BURLEIGH-TWR-snap-4Gtrailcam_jpg.rf.3WehESpX0Qzw7Q91a9NU.jpg",

    "1640149202-Wed-Dec-22_15_00_02-AEST-2021-BURLEIGH-TWR-snap-4Gtrailcam_jpg.rf.ls85n6g3Mrg5KoE9rfKw.jpg",

    "1640155505-Wed-Dec-22_16_45_05-AEST-2021-BURLEIGH-TWR-snap-4Gtrailcam_jpg.rf.osRyOFMWqcUQgxgfWbLo.jpg",

    "1640155503-Wed-Dec-22_16_45_03-AEST-2021-BURLEIGH-TWR-snap-4Gtrailcam_jpg.rf.622aUMZx7BgnBQE6c4Y5.jpg",

    "1640135702-Wed-Dec-22_11_15_02-AEST-2021-BURLEIGH-TWR-snap-4Gtrailcam_jpg.rf.7e47Phitp5U5GjN16rdq.jpg",

    "1640217602-Thu-Dec-23_10_00_02-AEST-2021-BURLEIGH-TWR-snap-4Gtrailcam_jpg.rf.1y4GGYLPw1jTry0zY4fQ.jpg",

    "1639953906-Mon-Dec-20_08_45_06-AEST-2021-BURLEIGH-TWR-snap-4Gtrailcam_jpg.rf.0dGyCV5KVtDrcRPFVPhs.jpg",

    "1640293205-Fri-Dec-24_07_00_05-AEST-2021-BURLEIGH-TWR-snap-4Gtrailcam_jpg.rf.ZaOpUQjD4cHVJMEGcoiO.jpg"
]

# ============================================================
# 5. FIND AVAILABLE REQUESTED IMAGES
# ============================================================

available = []

print("\n" + "=" * 70)
print("CHECKING REQUESTED IMAGES")
print("=" * 70)

for filename in requested_images:

    path = os.path.join(
        TEST_IMAGE_DIR,
        filename
    )

    if os.path.exists(path):

        available.append(path)

        print("✅", filename)

    else:

        print("❌ Missing:", filename)

print(
    "\nRequested images available:",
    len(available),
    "/ 10"
)

# ============================================================
# 6. AUTOMATICALLY FILL MISSING SLOTS
# ============================================================

all_test_images = sorted([
    os.path.join(TEST_IMAGE_DIR, f)
    for f in os.listdir(TEST_IMAGE_DIR)
    if f.lower().endswith(
        (".jpg", ".jpeg", ".png", ".bmp")
    )
])

used_names = set(
    os.path.basename(p)
    for p in available
)

remaining_images = [
    p for p in all_test_images
    if os.path.basename(p) not in used_names
]

needed = 10 - len(available)

if needed > 0:

    replacements = remaining_images[:needed]

    print("\n" + "=" * 70)
    print("REPLACING MISSING IMAGES")
    print("=" * 70)

    for p in replacements:

        print(
            "🔄 Replacement:",
            os.path.basename(p)
        )

        available.append(p)

# Make sure exactly 10
image_paths = available[:10]

print("\n" + "=" * 70)
print("FINAL 10 IMAGES")
print("=" * 70)

for i, p in enumerate(image_paths, 1):

    print(
        f"{i}. {os.path.basename(p)}"
    )

print(
    "\nTotal selected:",
    len(image_paths)
)

if len(image_paths) < 10:

    raise RuntimeError(
        "The test folder contains fewer than 10 images."
    )

# ============================================================
# 7. READ GROUND TRUTH LABELS
# ============================================================

def read_ground_truth(image_path):

    image_name = os.path.basename(
        image_path
    )

    label_name = (
        os.path.splitext(image_name)[0]
        + ".txt"
    )

    label_path = os.path.join(
        TEST_LABEL_DIR,
        label_name
    )

    if not os.path.exists(label_path):

        print(
            "⚠️ No label found:",
            label_name
        )

        return [], []

    image = Image.open(
        image_path
    ).convert("RGB")

    width, height = image.size

    boxes = []
    labels = []

    with open(label_path, "r") as f:

        for line in f:

            parts = line.strip().split()

            if len(parts) != 5:
                continue

            cls = int(parts[0])

            if cls not in [0, 1]:
                continue

            xc = float(parts[1])
            yc = float(parts[2])
            w = float(parts[3])
            h = float(parts[4])

            x1 = (xc - w / 2) * width
            y1 = (yc - h / 2) * height

            x2 = (xc + w / 2) * width
            y2 = (yc + h / 2) * height

            boxes.append([
                x1, y1, x2, y2
            ])

            labels.append(cls)

    return boxes, labels


# ============================================================
# 8. IOU
# ============================================================

def calculate_iou(box1, box2):

    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])

    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    intersection = (
        max(0, x2 - x1) *
        max(0, y2 - y1)
    )

    area1 = (
        max(0, box1[2] - box1[0]) *
        max(0, box1[3] - box1[1])
    )

    area2 = (
        max(0, box2[2] - box2[0]) *
        max(0, box2[3] - box2[1])
    )

    union = area1 + area2 - intersection

    if union <= 0:
        return 0

    return intersection / union


# ============================================================
# 9. RUN ALL 10 IMAGES
# ============================================================

results = []

all_gt_labels = []
all_pred_labels = []

total_tp = 0
total_fp = 0
total_fn = 0

print("\n")
print("=" * 70)
print("RUNNING MODEL ON 10 IMAGES")
print("=" * 70)

for image_number, image_path in enumerate(
    image_paths,
    start=1
):

    image = Image.open(
        image_path
    ).convert("RGB")

    image_tensor = F.to_tensor(
        image
    ).to(device)

    # --------------------------------------------------------
    # PREDICTION
    # --------------------------------------------------------

    with torch.no_grad():

        prediction = model(
            [image_tensor]
        )[0]

    boxes = prediction["boxes"].cpu().numpy()
    labels = prediction["labels"].cpu().numpy()
    scores = prediction["scores"].cpu().numpy()

    keep = scores >= CONFIDENCE_THRESHOLD

    boxes = boxes[keep]
    labels = labels[keep]
    scores = scores[keep]

    raw_labels = np.array([
        MODEL_TO_RAW.get(
            int(label),
            int(label)
        )
        for label in labels
    ])

    # --------------------------------------------------------
    # PREDICTED COUNTS
    # --------------------------------------------------------

    pred_land = int(
        np.sum(raw_labels == 0)
    )

    pred_water = int(
        np.sum(raw_labels == 1)
    )

    pred_total = (
        pred_land +
        pred_water
    )

    # --------------------------------------------------------
    # GROUND TRUTH
    # --------------------------------------------------------

    gt_boxes, gt_labels = read_ground_truth(
        image_path
    )

    gt_land = sum(
        1 for x in gt_labels
        if x == 0
    )

    gt_water = sum(
        1 for x in gt_labels
        if x == 1
    )

    gt_total = len(gt_labels)

    # --------------------------------------------------------
    # MATCH PREDICTIONS TO GROUND TRUTH
    # --------------------------------------------------------

    matched_gt = set()

    tp = 0
    fp = 0

    for pred_box, pred_label in zip(
        boxes,
        raw_labels
    ):

        best_iou = 0
        best_gt = -1

        for j, gt_box in enumerate(
            gt_boxes
        ):

            if j in matched_gt:
                continue

            iou = calculate_iou(
                pred_box,
                gt_box
            )

            if iou > best_iou:

                best_iou = iou
                best_gt = j

        if best_iou >= IOU_THRESHOLD:

            matched_gt.add(best_gt)

            tp += 1

            all_gt_labels.append(
                gt_labels[best_gt]
            )

            all_pred_labels.append(
                int(pred_label)
            )

        else:

            fp += 1

    fn = len(gt_boxes) - len(matched_gt)

    total_tp += tp
    total_fp += fp
    total_fn += fn

    # --------------------------------------------------------
    # DISPLAY IMAGE
    # --------------------------------------------------------

    fig, ax = plt.subplots(
        figsize=(13, 8)
    )

    ax.imshow(image)

    for box, raw_label, score in zip(
        boxes,
        raw_labels,
        scores
    ):

        x1, y1, x2, y2 = box

        rect = patches.Rectangle(
            (x1, y1),
            x2 - x1,
            y2 - y1,
            linewidth=2,
            edgecolor="lime",
            facecolor="none"
        )

        ax.add_patch(rect)

        ax.text(
            x1,
            max(y1 - 5, 5),
            f"{raw_label}:{score:.2f}",
            color="lime",
            fontsize=9,
            weight="bold"
        )

    # --------------------------------------------------------
    # COUNT PANEL
    # --------------------------------------------------------

    panel_text = (
        f"Land (0): {pred_land}    "
        f"Water (1): {pred_water}    "
        f"Total: {pred_total}\n"
        f"Ground Truth: {gt_total}"
    )

    ax.text(
        0.02,
        0.98,
        panel_text,
        transform=ax.transAxes,
        fontsize=11,
        weight="bold",
        color="white",
        verticalalignment="top",
        bbox=dict(
            boxstyle="round,pad=0.35",
            facecolor="black",
            alpha=0.65,
            edgecolor="none"
        )
    )

    ax.set_title(
        f"Image {image_number}/10",
        fontsize=13,
        weight="bold"
    )

    ax.axis("off")

    plt.tight_layout()

    plt.show()

    # --------------------------------------------------------
    # PRINT
    # --------------------------------------------------------

    print(
        f"{image_number}. "
        f"{os.path.basename(image_path)}"
    )

    print(
        f"   Prediction → "
        f"Land={pred_land}, "
        f"Water={pred_water}, "
        f"Total={pred_total}"
    )

    print(
        f"   Ground Truth → "
        f"Land={gt_land}, "
        f"Water={gt_water}, "
        f"Total={gt_total}"
    )

    print(
        f"   TP={tp}, FP={fp}, FN={fn}"
    )

    print()

    results.append({

        "Image": image_number,

        "Filename":
            os.path.basename(image_path),

        "Ground Truth Land":
            gt_land,

        "Predicted Land":
            pred_land,

        "Ground Truth Water":
            gt_water,

        "Predicted Water":
            pred_water,

        "Ground Truth Total":
            gt_total,

        "Predicted Total":
            pred_total,

        "Error":
            pred_total - gt_total,

        "Absolute Error":
            abs(pred_total - gt_total),

        "TP":
            tp,

        "FP":
            fp,

        "FN":
            fn
    })


# ============================================================
# 10. RESULTS TABLE
# ============================================================

df = pd.DataFrame(results)

print("\n")
print("=" * 90)
print("FINAL RESULTS FOR 10 IMAGES")
print("=" * 90)

display(df)

# ============================================================
# 11. OVERALL METRICS
# ============================================================

precision_det = (
    total_tp /
    (total_tp + total_fp)
    if total_tp + total_fp > 0
    else 0
)

recall_det = (
    total_tp /
    (total_tp + total_fn)
    if total_tp + total_fn > 0
    else 0
)

f1_det = (
    2 * precision_det * recall_det /
    (precision_det + recall_det)
    if precision_det + recall_det > 0
    else 0
)

print("\n")
print("=" * 60)
print("OVERALL DETECTION METRICS")
print("=" * 60)

print("TP =", total_tp)
print("FP =", total_fp)
print("FN =", total_fn)

print(
    f"Precision = {precision_det:.4f}"
)

print(
    f"Recall    = {recall_det:.4f}"
)

print(
    f"F1 Score  = {f1_det:.4f}"
)

# ============================================================
# 12. COUNTING METRICS
# ============================================================

print("\n")
print("=" * 60)
print("COUNTING PERFORMANCE")
print("=" * 60)

print(
    "Total Ground Truth:",
    df["Ground Truth Total"].sum()
)

print(
    "Total Predicted:",
    df["Predicted Total"].sum()
)

print(
    "Mean Absolute Error:",
    round(
        df["Absolute Error"].mean(),
        3
    )
)

print(
    "Total Absolute Error:",
    df["Absolute Error"].sum()
)

# ============================================================
# 13. CONFUSION MATRIX
# ============================================================

if len(all_gt_labels) > 0:

    y_true = np.array(
        all_gt_labels
    )

    y_pred = np.array(
        all_pred_labels
    )

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    print("\n")
    print("=" * 60)
    print("CONFUSION MATRIX")
    print("=" * 60)

    print(cm)

    fig, ax = plt.subplots(
        figsize=(7, 6)
    )

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=[
            "0: Land",
            "1: Water"
        ]
    )

    disp.plot(
        ax=ax,
        cmap="Blues"
    )

    plt.title(
        "Confusion Matrix — 10 Images"
    )

    plt.tight_layout()

    plt.show()

    # --------------------------------------------------------
    # CLASSIFICATION METRICS
    # --------------------------------------------------------

    print(
        "Classification Precision:",
        round(
            precision_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            ),
            4
        )
    )

    print(
        "Classification Recall:",
        round(
            recall_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            ),
            4
        )
    )

    print(
        "Classification F1:",
        round(
            f1_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            ),
            4
        )
    )

# ============================================================
# 14. GRAPH — GROUND TRUTH VS PREDICTED TOTAL
# ============================================================

x = np.arange(
    1,
    len(df) + 1
)

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    x,
    df["Ground Truth Total"],
    marker="o",
    linewidth=2,
    label="Ground Truth"
)

plt.plot(
    x,
    df["Predicted Total"],
    marker="s",
    linewidth=2,
    label="Predicted"
)

plt.xlabel("Image Number")
plt.ylabel("Number of People")

plt.title(
    "Ground Truth vs Predicted People Count"
)

plt.xticks(x)

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()

# ============================================================
# 15. GRAPH — LAND
# ============================================================

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    x,
    df["Ground Truth Land"],
    marker="o",
    linewidth=2,
    label="Ground Truth Land (0)"
)

plt.plot(
    x,
    df["Predicted Land"],
    marker="s",
    linewidth=2,
    label="Predicted Land (0)"
)

plt.xlabel("Image Number")
plt.ylabel("People")

plt.title(
    "Land People — Ground Truth vs Prediction"
)

plt.xticks(x)

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()

# ============================================================
# 16. GRAPH — WATER
# ============================================================

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    x,
    df["Ground Truth Water"],
    marker="o",
    linewidth=2,
    label="Ground Truth Water (1)"
)

plt.plot(
    x,
    df["Predicted Water"],
    marker="s",
    linewidth=2,
    label="Predicted Water (1)"
)

plt.xlabel("Image Number")
plt.ylabel("People")

plt.title(
    "Water People — Ground Truth vs Prediction"
)

plt.xticks(x)

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()

# ============================================================
# 17. GRAPH — COUNT ERROR
# ============================================================

plt.figure(
    figsize=(12, 6)
)

plt.bar(
    x,
    df["Error"]
)

plt.axhline(
    0,
    linewidth=1
)

plt.xlabel("Image Number")
plt.ylabel("Prediction Error")

plt.title(
    "People Count Error"
)

plt.xticks(x)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()

# ============================================================
# 18. GRAPH — TP / FP / FN
# ============================================================

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    x,
    df["TP"],
    marker="o",
    linewidth=2,
    label="TP"
)

plt.plot(
    x,
    df["FP"],
    marker="s",
    linewidth=2,
    label="FP"
)

plt.plot(
    x,
    df["FN"],
    marker="^",
    linewidth=2,
    label="FN"
)

plt.xlabel("Image Number")
plt.ylabel("Detections")

plt.title(
    "TP / FP / FN per Image"
)

plt.xticks(x)

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()

# ============================================================
# 19. SAVE RESULTS
# ============================================================

RESULTS_DIR = os.path.join(
    DATASET_PATH,
    "results"
)

os.makedirs(
    RESULTS_DIR,
    exist_ok=True
)

csv_path = os.path.join(
    RESULTS_DIR,
    "10_images_results.csv"
)

df.to_csv(
    csv_path,
    index=False
)

print("\n")
print("=" * 70)
print("✅ COMPLETE")
print("=" * 70)

print(
    "Results CSV saved at:"
)

print(csv_path)

print("\nThe 10 images were successfully evaluated.")
```
